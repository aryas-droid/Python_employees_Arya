Tasks
Your program should perform the following operations:
1. Display the total number of employees.
2. Display the average, highest, and lowest salary.
3. Display the names of employees whose salary is greater than ₹65,000 and whose
performance rating is 4.5 or above.
4. Create a dictionary with Employee ID as the key and Employee Name as the value.
5. Create a list containing all employee names, sort it alphabetically, reverse it, and display
the first three names.
6. Store all Employee IDs in a tuple and display the maximum and minimum Employee ID.
7. Create a set of unique departments and display the department names.
8. Accept an Employee ID from the user and display the employee details if found;
otherwise display "Employee Not Found".

```python
employees = [
 {"id":101, "name":"Alice", "department":"Data", "salary":65000, "rating":4.5},
 {"id":102, "name":"Bob", "department":"AI", "salary":72000, "rating":4.8},
 {"id":103, "name":"Charlie", "department":"Web", "salary":58000, "rating":3.9},
 {"id":104, "name":"David", "department":"Data", "salary":61000, "rating":4.2},
 {"id":105, "name":"Eva", "department":"AI", "salary":75000, "rating":4.9},
 {"id":106, "name":"Frank", "department":"Cloud", "salary":68000, "rating":4.1},
 {"id":107, "name":"Grace", "department":"Web", "salary":59000, "rating":3.8},
 {"id":108, "name":"Helen", "department":"Data", "salary":70000, "rating":4.6}
]
print("Total no of employees: ",len(employees))
average_salary=sum(employee["salary"] for employee in employees)/len(employees)
print("Average salary: ",average_salary)
highest_salary=max(employee["salary"] for employee in employees)
print("Highest salary: ",highest_salary)
lowest_salary=min(employee["salary"] for employee in employees)
print("Lowest salary: ",lowest_salary)
for employee in employees:
  if employee["salary"]>65000 and employee["rating"]>=4.5:
    print(employee)
id_name=dict((employee["id"], employee["name"]) for employee in employees)
print(id_name)
result=sorted([e["name"] for e in employees],reverse=True)[:3]
print(result)




```

```python
# Store all Employee IDs in a tuple and display the maximum and minimum Employee ID
employee_ids = tuple(employee["id"] for employee in employees)
print("Employee IDs:", employee_ids)
print("Maximum Employee ID:", max(employee_ids))
print("Minimum Employee ID:", min(employee_ids))

#Create a set of unique departments and display the department names
depart= set(employee["department"] for employee in employees)
print("Unique departments:", depart)

#Accept an Employee ID from the user and display the employee details if found; otherwise display "Employee Not Found".
employee_id = int(input("Enter employee ID: "))
employee_found=False
for employee in employees:
  if employee["id"]==employee_id:
    print("Employee details: ",employee)
    employee_found=True
    break
if not employee_found:
  print("Employee not found")
```
